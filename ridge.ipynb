```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Advertising_cleaned.csv')
df
```

```python
X = df[['total_spend']]
y=df['sales']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

# GridSearchCV

```python
from sklearn.model_selection import GridSearchCV 

from sklearn.linear_model import Ridge 
estimator = Ridge()

param_grid = {'alpha':list(range(1,100))}

r_gs = GridSearchCV(estimator,param_grid,scoring='r2',cv=5)
r_gs.fit(X_train,y_train)
r_gs.best_estimator_
```

```python
from sklearn.linear_model import Ridge
model = Ridge(alpha=99)
model.fit(X_train,y_train)
```

```python
print('COeff:',model.coef_)
print('Intercept:',model.intercept_)
```

```python
y_pred_train = model.predict(X_train)
from sklearn.metrics import r2_score 
print('Train R2:',r2_score(y_train,y_pred_train))
```

```python
ypred_test = model.predict(X_test)
print('Test r2:',r2_score(y_test,ypred_test))
```

```python
from sklearn.model_selection import GridSearchCV 

from sklearn.linear_model import Lasso
estimator = Lasso()

param_grid = {'alpha':list(range(1,100))}

lasso_gs = GridSearchCV(estimator,param_grid,scoring='r2',cv=5)
lasso_gs.fit(X_train,y_train)
lasso_gs.best_estimator_
```

```python
from sklearn.linear_model import Lasso 
model = Lasso(alpha=99)
model.fit(X_train,y_train)
```

```python
print('COeff:',model.coef_)
print('Intercept:',model.intercept_)
```

```python
y_pred_train = model.predict(X_train)
from sklearn.metrics import r2_score 
print('Train R2:',r2_score(y_train,y_pred_train))
```

```python
ypred_test = model.predict(X_test)
print('Test r2:',r2_score(y_test,ypred_test))
```

```python
from sklearn.model_selection import GridSearchCV 

from sklearn.linear_model import ElasticNet
estimator = ElasticNet()

param_grid = {'alpha':list(range(1,70))}

el_gs = GridSearchCV(estimator,param_grid,scoring='r2',cv=5)
el_gs.fit(X_train,y_train)
el_gs.best_estimator_
```

```python
from sklearn.linear_model import ElasticNet
model = ElasticNet(alpha=59,l1_ratio=0.5)
model.fit(X_train,y_train)
```

```python
print('COeff:',model.coef_)
print('Intercept:',model.intercept_)
```

```python
y_pred_train = model.predict(X_train)
from sklearn.metrics import r2_score 
print('Train R2:',r2_score(y_train,y_pred_train))
```

```python
ypred_test = model.predict(X_test)
print('Test r2:',r2_score(y_test,ypred_test))
```

# polynomial Regression

```python
from sklearn.model_selection import train_test_split 
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=77)
```

```python
from sklearn.preprocessing import PolynomialFeatures 
polynomial_converter = PolynomialFeatures(degree=3,include_bias=False)

X_train_poly = pd.DataFrame(polynomial_converter.fit_transform(X_train))
X_test_poly = pd.DataFrame(polynomial_converter.transform(X_test))
```

```python
X_train_poly
```

```python
X_test_poly
```

```python
from sklearn.linear_model import LinearRegression 
model = LinearRegression()
model.fit(X_train_poly,y_train)
```

```python
print('coeffs:',model.coef_)
print('Intercept:',model.intercept_)
```

```python
ypred_train = model.predict(X_train_poly)
from sklearn.metrics import r2_score 
print('Train r2:',r2_score(y_train,ypred_train))
```

```python
ypred_test = model.predict(X_test_poly)
print('Test R2:',r2_score(y_test,ypred_test))
```
